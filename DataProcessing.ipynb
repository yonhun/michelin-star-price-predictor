```python
import pandas as pd
```

```python
df = pd.read_csv('michelin_my_maps.csv')
df
```

```python
# 'location' 컬럼을 쉼표 기준으로 분리하여 'city'와 'region' 컬럼 생성
df[['City', 'Region']] = df['Location'].str.split(',', expand=True)

# 'city' 컬럼의 앞뒤 공백 제거
df['City'] = df['City'].str.strip()

# 'region' 컬럼의 앞뒤 공백 제거
df['Region'] = df['Region'].str.strip()

df.head()
```

```python
df['Year'] =2024
df.head()
```

```python
df.drop('Address', axis=1, inplace=True)
df.drop('Location', axis=1, inplace=True)
df.drop('Name', axis=1, inplace=True)
df.drop('WebsiteUrl', axis=1, inplace=True)
df.drop('Url', axis=1, inplace=True)
df.drop('GreenStar', axis=1, inplace=True)
df.drop('Description', axis=1, inplace=True)
df.drop('PhoneNumber', axis=1, inplace=True)
df.drop('FacilitiesAndServices', axis=1, inplace=True)
df.head()
```

```python
df['Award'].unique()
```

```python
# 원하는 값의 리스트를 생성합니다
desired_awards = ['3 Stars', '2 Stars', '1 Star']

# 'Award' 칼럼의 값이 원하는 값들인 행만 선택합니다
df = df[df['Award'].isin(desired_awards)]

df
```

```python
df['Award'] = df['Award'].str.extract('(\d+)', expand=False)

df.head()
```

```python
df = df.rename(columns={'Award': 'Stars'})
df.head()
```

```python
df['Price'].unique()
```

```python
df['Price'] = df['Price'].str.len()
df['Price'].unique()
```

```python
new_column_order = ['Year', 'Latitude', 'Longitude', 'City', 'Region', 'Cuisine', 'Price', 'Stars']
df = df[new_column_order]
df.head()
```

```python
df_3Stars = df[df['Stars'] == '3']
df_2Stars = df[df['Stars'] == '2']
df_1Stars = df[df['Stars'] == '1']
```

```python
df_3Stars
```

```python
df_2Stars.reset_index(drop=True, inplace=True)
df_2Stars
```

```python
df_1Stars.reset_index(drop=True, inplace=True)
df_1Stars
```

```python
df_1Stars.to_csv('one-star-michelin-restaurants-2024.csv', index=False)
df_2Stars.to_csv('two-stars-michelin-restaurants-2024.csv', index=False)
df_3Stars.to_csv('three-stars-michelin-restaurants-2024.csv', index=False)
```
